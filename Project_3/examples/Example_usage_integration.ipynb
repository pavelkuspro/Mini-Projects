# Goal
The purpose of this notebook is to demonstrate the application of the **explicit Euler method** and the **explicit 4th-order Runge-Kutta method (RK4)** to solve ordinary differential equations (ODEs) of the form:

$$
\frac{dy}{dx} = f(x, y(x)),
$$  

with given initial values \((x_0, y_0)\).

```python
%load_ext autoreload
%autoreload 2

import numpy as np
import matplotlib.pyplot as plt
from numerical_methods import explicit_euler, explicit_runge_kutta4
```

## Illustrative Example

```python
def f(x,y):
    return x * np.sin(x)
```

```python
x_list_ee, y_list_ee = explicit_euler(f = f, x0 = 0, y0 = 0, h = 0.001, n = 10000)
x_list_erk, y_list_erk = explicit_runge_kutta4(f = f, x0 = 0, y0 = 0, h = 0.001, n = 10000)
```

```python
plt.figure(figsize = (10,4))
plt.plot(x_list_ee[::100], y_list_ee[::100], "*-", color = "blue", label = r"$y_{EE}(x)$")
plt.plot(x_list_erk[::125], y_list_erk[::125], "v--", color = "red", label = r"$y_{RG}(x)$", alpha = 0.3)
plt.xlabel("x")
plt.ylabel("y")
plt.legend(loc = "upper left")
plt.show()
```
